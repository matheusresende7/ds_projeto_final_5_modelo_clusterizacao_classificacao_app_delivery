## 01.3. União entre tabelas

### 01.3.1. Habilitando a exibição de todas as colunas

```python
pd.set_option('display.max_columns', None)
```

### 01.3.2. Entendendo as tabelas e seus tipos de dados

```python
dim_customers = pd.read_csv(consts.DIM_CUSTOMERS_RAW)

dim_calender = pd.read_csv(consts.DIM_CALENDER_RAW, sep = ';')
```

### 01.3.3. Unindo as tabelas

```python
dim_customers['Dt_Customer'] = pd.to_datetime(dim_customers['Dt_Customer'])

dim_calender['Data'] = pd.to_datetime(dim_calender['Data'])

dataset_raw = dim_customers.merge(

    dim_calender[['Data', 'Mês', 'Trimestre']],

    left_on = 'Dt_Customer',

    right_on = 'Data',

    how = 'left'
    
)

dataset_raw = dataset_raw.drop(['Data'], axis = 1)

dataset_raw = dataset_raw.rename(columns = {
    'Mês': 'Dt_Customer_Month',
    'Trimestre': 'Dt_Customer_Quarter'
})
```

```python
dataset_raw.head(5)
```

```python
dataset_raw.info()
```

### 01.3.4. Salvando a tabela unificada

```python
dataset_raw.to_csv(

    consts.DATASET_RAW_PANDAS,

    index = False

)
```
